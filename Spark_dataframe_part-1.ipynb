```python
from pyspark.sql import functions as F
```

```python
# Reading a table from catalog.
cust_df = spark.table("samples.bakehouse.sales_customers")
cust_df.display()
```

```python
cust_df.show()
```

```python
# Reading the table data using sql query
cust_df = spark.sql("select customerID, first_name, last_name, city, country, gender from samples.bakehouse.sales_customers")
cust_df.display()
```

```python
# selecting the data from a table
result_df = (
    cust_df
    .select(
        "customerID",
        "first_name",
        "last_name",
        "city",
        "country"
    )    
)
result_df.display()

```

```python
result_df = (
    cust_df
    .select(
        cust_df["customerID"],
        "first_name",
        "last_name",
        "city",
        "country"
        "gender"
    )    
)
result_df.display()

```

```python
result_df = (
    cust_df
    .select(
        cust_df["customerID"],
        cust_df["first_name"],
        cust_df["last_name"],
        cust_df["city"],
        cust_df["country"],
        cust_df["gender"]
    )    
)
result_df.display()
```

```python
result_df = (
    cust_df
    .select(
        cust_df.customerID,
        cust_df.first_name,
        cust_df.last_name,
        cust_df.city,
        cust_df.country
    )    
)
result_df.display()
```

```python
result_df = (
    cust_df
    .select(
        F.col("customerID"),
        F.col("first_name"),
        F.col("last_name"),
        F.col("city"),
        F.col("country"),
        F.col("gender")
    )    
)
result_df.display()
```

```python
# Filtering data from dataframe
result_df = (
    cust_df
    .filter(
        (F.col("gender") == "male") &
        (F.col("country") == "USA")
    )    
)
result_df.display()
```
