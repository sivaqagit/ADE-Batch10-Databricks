```python
custo_df = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", True)
    .load("/Volumes/rr_qc_my_workspace/qc_db/quant_files/Customer-Churn-Records.csv")
)
custo_df.display()
```

```python
custo_df = (
    custo_df
    .withColumn("RowNumber", F.col("RowNumber").cast("integer"))
    .withColumn("CustomerId", F.col("CustomerId").cast("integer"))
    
)
custo_df.display()
```

```python
custo_df = (
    custo_df
    .select(
        F.col("RowNumber").cast("int"),
        F.col("CustomerId").cast("int")  
    )
)
custo_df.display()
```

```python
# DAG Direct Acyclic Graph - A graph with no cycles (spark is a lazy evolution system)
#Transformations (read, filter, withColumn, select) are lazy, they are not executed until an action is called.
# actions(collect, count, display, write,take, show, saveAsTextFile, foreach, foreachPartition)

```

```python
old_columns = custo_df.columns
new_columns = [col.lower() for col in old_columns]
custo_df = (
    custo_df
    .select(
        *[F.col(ocol).alias(ncol) for ncol, ocol in zip(new_columns, old_columns)]
    )
)
custo_df.display()
```
